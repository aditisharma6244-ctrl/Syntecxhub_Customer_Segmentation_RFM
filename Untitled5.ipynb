```python

import pandas as pd

data = {
    "CustomerID": [101, 101, 102, 102, 103, 103, 104, 105, 105, 106,
                   107, 107, 108, 109, 110],
    "InvoiceDate": [
        "2026-09-10", "2026-09-20", "2026-08-15", "2026-09-05",
        "2026-07-10", "2026-08-20", "2026-09-25", "2026-06-15",
        "2026-09-01", "2026-09-18", "2026-08-25", "2026-09-12",
        "2026-09-28", "2026-07-20", "2026-09-22"
    ],
    "Amount": [500, 800, 300, 450, 1200, 700, 250, 1500,
               600, 400, 350, 900, 200, 1000, 550]
}

df = pd.DataFrame(data)

print("Customer Transaction Dataset:")
display(df)
print("Dataset shape:", df.shape)
```

```python

print("Dataset Information:")
df.info()

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDuplicate Rows:", df.duplicated().sum())
```

```python

# Convert InvoiceDate into datetime format
df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])

# Check and remove duplicate rows
df = df.drop_duplicates()

# Check for missing values
print("Missing values before cleaning:")
print(df.isnull().sum())

# Remove rows with missing essential values
df = df.dropna(subset=["CustomerID", "InvoiceDate", "Amount"])

# Display cleaned dataset
print("\nCleaned Dataset:")
display(df)

print("Total transactions after cleaning:", len(df))
```

```python

# Set the analysis date
analysis_date = pd.Timestamp("2026-10-01")

# Calculate RFM metrics for each customer
rfm = df.groupby("CustomerID").agg(
    Recency=("InvoiceDate", lambda x: (analysis_date - x.max()).days),
    Frequency=("InvoiceDate", "count"),
    Monetary=("Amount", "sum")
).reset_index()

print("RFM Analysis Results:")
display(rfm)
```

```python

# Assign scores based on customer behavior

rfm["R_Score"] = rfm["Recency"].apply(
    lambda x: 3 if x <= 10 else (2 if x <= 30 else 1)
)

rfm["F_Score"] = rfm["Frequency"].apply(
    lambda x: 3 if x >= 3 else (2 if x == 2 else 1)
)

rfm["M_Score"] = rfm["Monetary"].apply(
    lambda x: 3 if x >= 1500 else (2 if x >= 700 else 1)
)

# Classify customers into segments
def segment_customer(row):
    if row["R_Score"] >= 2 and row["F_Score"] >= 2 and row["M_Score"] >= 2:
        return "Loyal Customers"
    elif row["R_Score"] == 1:
        return "At Risk"
    elif row["R_Score"] == 3 and row["F_Score"] == 1:
        return "New Customers"
    else:
        return "Regular Customers"

rfm["Segment"] = rfm.apply(segment_customer, axis=1)

print("Customer Segmentation Results:")
display(rfm)
```

```python

import matplotlib.pyplot as plt

segment_counts = rfm["Segment"].value_counts()

segment_counts.plot(kind="bar", figsize=(8, 5))

plt.title("Customer Segmentation using RFM Analysis")
plt.xlabel("Customer Segment")
plt.ylabel("Number of Customers")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python

# Marketing recommendations for each customer segment

recommendations = {
    "Loyal Customers": "Offer loyalty points, rewards, and exclusive discounts.",
    "New Customers": "Provide a welcome offer to encourage another purchase.",
    "At Risk": "Send re-engagement emails and special discounts.",
    "Regular Customers": "Recommend relevant products and offer bundle deals."
}

rfm["Marketing_Recommendation"] = rfm["Segment"].map(recommendations)

print("Customer Segmentation with Marketing Recommendations:")
display(rfm)
```

```python

# Save the RFM analysis results
rfm.to_csv("customer_segmentation_results.csv", index=False)

# Save the cleaned transaction dataset
df.to_csv("cleaned_customer_transactions.csv", index=False)

print("Project files saved successfully!")
print("1. customer_segmentation_results.csv")
print("2. cleaned_customer_transactions.csv")
```
